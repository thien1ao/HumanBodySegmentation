# Human Body Segmentation on Meshes with PyTorch3D

**Goal:** given a 3D triangle mesh of a human, predict for every triangle which body part it belongs to
(head, hand, forearm, upper arm, torso, thigh, shin, foot).

**Plan of this notebook**

1. Load the `human_seg` dataset (from the MeshCNN paper) and understand its label format.
2. Visualize the ground-truth segmentation with the PyTorch3D renderer.
3. Notice a big problem — the meshes are **oriented differently** depending on where they come from — and
   solve it with **intrinsic** (rotation-invariant) geometry: the **Laplace–Beltrami operator**, its
   **eigenfunctions**, and the **Heat Kernel Signature (HKS)**.
4. Build a small **DiffusionNet-style** neural network that passes information along the surface by
   simulating heat diffusion.
5. Train it, evaluate it (accuracy, IoU, confusion matrix) and look at the predictions.

Everything runs on a single GPU in a few minutes (CPU also works, just slower).

## 1. Setup

* **PyTorch** — tensors, autograd, the neural network.
* **PyTorch3D** — loading `.obj` files (`load_obj`), the `Meshes` structure and the differentiable
  renderer we use for pictures.
* **matplotlib** — plots and showing rendered images.

We pick the GPU if one is available. Seeds are fixed so the results are reproducible.

```python
import os, glob, math, time, random, warnings
from collections import Counter, defaultdict

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt

from pytorch3d.io import load_obj
from pytorch3d.structures import Meshes
from pytorch3d.renderer import (
    FoVPerspectiveCameras, MeshRasterizer, MeshRenderer, HardPhongShader,
    PointLights, RasterizationSettings, TexturesVertex, look_at_view_transform,
)

warnings.filterwarnings("ignore", message="No mtl file")   # the .obj files have no materials

device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
print("device:", device)

torch.manual_seed(0); np.random.seed(0); random.seed(0)

DATA_ROOT = "human_seg"          # folder with train/ test/ seg/ sseg/
```

## 2. The dataset

`human_seg` contains human meshes collected from several sources
(SCAPE, FAUST, MIT animation, Adobe Fuse for training; SHREC for testing).
Every mesh was simplified to roughly **750 vertices / 1500 triangles**.

Folder layout:

| folder | contents |
|---|---|
| `train/*.obj`, `test/*.obj` | the meshes |
| `seg/*.eseg` | one **hard label per edge** (an integer 1–8) |
| `sseg/*.seseg` | **soft labels per edge** (8 probabilities), used by MeshCNN for evaluation |

The file name prefix (before `__`) tells us which source the mesh came from. Let's count.

```python
train_files = sorted(glob.glob(f"{DATA_ROOT}/train/*.obj"))
test_files  = sorted(glob.glob(f"{DATA_ROOT}/test/*.obj"))
print(f"train meshes: {len(train_files)}   test meshes: {len(test_files)}")

source = lambda p: os.path.basename(p).split("__")[0]
print("train sources:", dict(Counter(map(source, train_files))))
print("test sources: ", dict(Counter(map(source, test_files))))

CLASS_NAMES = ["head", "hand", "forearm", "upper arm", "torso", "thigh", "shin", "foot"]
NUM_CLASSES = len(CLASS_NAMES)
```

## 3. From edge labels to face labels

The labels are stored **per edge**, because MeshCNN works on edges. A mesh file only stores vertices and
faces, so the edge list has to be *rebuilt*, and it must be rebuilt **in exactly the same order** MeshCNN used,
otherwise label *i* would be attached to the wrong edge.

MeshCNN's order: walk over the faces in file order; for each face `(a, b, c)` visit the edges
`(a,b)`, `(b,c)`, `(c,a)`; an edge gets the next free index the **first** time it is seen.

We want **one label per face** (triangle), which is more natural for rendering and for our network.
A triangle has 3 edges, so we take a **majority vote** of the 3 edge labels (if all three differ we take the first).

We also **normalize** every mesh: move it to the origin and scale it to **total surface area = 1**.
Different sources use different units (meters, centimeters…); after this step all humans have the same "size",
which matters later because heat diffusion depends on scale.

```python
def meshcnn_edge_ids(faces):
    # dict {(min_vertex, max_vertex): edge_index} in MeshCNN's order
    edge_id = {}
    for f in faces.tolist():
        for k in range(3):
            e = tuple(sorted((f[k], f[(k + 1) % 3])))
            if e not in edge_id:
                edge_id[e] = len(edge_id)
    return edge_id


def face_labels_from_edges(faces, edge_labels):
    edge_id = meshcnn_edge_ids(faces)
    assert len(edge_id) == len(edge_labels), "edge count does not match label count"
    out = []
    for f in faces.tolist():
        labs = [edge_labels[edge_id[tuple(sorted((f[k], f[(k + 1) % 3])))]] for k in range(3)]
        out.append(Counter(labs).most_common(1)[0][0])
    return torch.tensor(out) - 1          # file labels are 1..8 -> we use 0..7


def face_areas(V, Fc):
    return 0.5 * torch.linalg.cross(V[Fc[:, 1]] - V[Fc[:, 0]], V[Fc[:, 2]] - V[Fc[:, 0]]).norm(dim=1)


def load_example(obj_path):
    verts, faces, _ = load_obj(obj_path, load_textures=False)
    V, Fc = verts.double(), faces.verts_idx.long()
    name = os.path.splitext(os.path.basename(obj_path))[0]
    edge_labels = np.loadtxt(f"{DATA_ROOT}/seg/{name}.eseg", dtype=np.int64)
    labels = face_labels_from_edges(Fc, edge_labels)

    A = face_areas(V, Fc)
    centroid = ((V[Fc].mean(1)) * A[:, None]).sum(0) / A.sum()   # area-weighted center
    V = (V - centroid) / A.sum().sqrt()                          # total area becomes 1
    return {"name": name, "source": source(obj_path), "V": V, "F": Fc, "labels": labels}


ex = load_example(train_files[0])
print(ex["name"], "| vertices:", len(ex["V"]), "| faces:", len(ex["F"]))
print("surface area after normalization:", face_areas(ex["V"], ex["F"]).sum().item())
print("faces per class:", torch.bincount(ex["labels"], minlength=NUM_CLASSES).tolist())
```

## 4. Rendering meshes with PyTorch3D

To see the segmentation we color each triangle by its class and render the mesh.

A few details worth explaining:

* **Flat colors per face.** PyTorch3D's simplest texture, `TexturesVertex`, stores a color per *vertex* and
  blends colors inside triangles. A vertex on a part boundary is shared by triangles of two classes, so the border
  would be blurry. Trick: give every triangle its **own copy of its 3 vertices** ("un-sharing"), then all three
  copies get the face color and borders are crisp.
* **Camera.** `look_at_view_transform(dist, elev, azim)` places a camera on a sphere around the origin looking at it.
  Several azimuths give us views from several sides in one batched render.
* **Display orientation.** As we'll see in the next section, the meshes are not all upright. *For pictures only*
  we rotate each mesh so that the axis from the feet to the head points up — we can do that because we know the labels.
  The network will never use this.

```python
PALETTE = torch.tensor([
    [0.90, 0.30, 0.30], [0.30, 0.70, 0.30], [0.30, 0.45, 0.90], [0.95, 0.75, 0.20],
    [0.70, 0.35, 0.85], [0.20, 0.80, 0.80], [0.95, 0.55, 0.15], [0.60, 0.60, 0.60],
])


def upright_rotation(V, Fc, labels):
    # Rotation that makes feet->head point along +y and the torso's thin direction along z (display only).
    C = V[Fc].mean(1)
    up = C[labels == 0].mean(0) - C[labels == 7].mean(0)
    up = up / up.norm()
    torso = C[labels == 4] - C[labels == 4].mean(0)
    torso = torso - (torso @ up)[:, None] * up            # remove the vertical component
    _, _, Vt = torch.linalg.svd(torso, full_matrices=False)
    side = Vt[0] - (Vt[0] @ up) * up; side = side / side.norm()   # shoulder-to-shoulder direction
    front = torch.linalg.cross(side, up)
    return torch.stack([side, up, front])                  # rows = new x, y, z axes


def render_face_colors(V, Fc, face_rgb, n_views=4, size=320, dist=1.6, R_disp=None):
    # Render a mesh with one RGB color per face, from n_views azimuths. Returns (n_views,H,W,3).
    V = V.float()
    if R_disp is not None:
        V = V @ R_disp.float().T
    v = V[Fc.reshape(-1)]                                   # un-share vertices
    f = torch.arange(len(v)).reshape(-1, 3)
    col = face_rgb.float().repeat_interleave(3, dim=0)
    mesh = Meshes([v.to(device)], [f.to(device)], textures=TexturesVertex([col.to(device)]))

    azim = torch.linspace(0, 360, n_views + 1)[:-1]
    R, T = look_at_view_transform(dist=dist, elev=10, azim=azim)
    cams = FoVPerspectiveCameras(device=device, R=R, T=T)
    lights = PointLights(device=device, location=[[0.0, 2.0, 3.0]], ambient_color=[[0.55] * 3],
                         diffuse_color=[[0.45] * 3], specular_color=[[0.05] * 3])
    renderer = MeshRenderer(
        MeshRasterizer(cameras=cams, raster_settings=RasterizationSettings(image_size=size, faces_per_pixel=1)),
        HardPhongShader(device=device, cameras=cams, lights=lights))
    return renderer(mesh.extend(n_views))[..., :3].clamp(0, 1).cpu().numpy()


def show_row(images, titles=None, suptitle=None, legend=False):
    fig, axes = plt.subplots(1, len(images), figsize=(3 * len(images), 3.3))
    for i, (ax, im) in enumerate(zip(np.atleast_1d(axes), images)):
        ax.imshow(im); ax.axis("off")
        if titles: ax.set_title(titles[i], fontsize=9)
    if legend:
        from matplotlib.patches import Patch
        fig.legend(handles=[Patch(color=PALETTE[i].tolist(), label=n) for i, n in enumerate(CLASS_NAMES)],
                   loc="lower center", ncol=8, fontsize=8, frameon=False)
    if suptitle: fig.suptitle(suptitle, fontsize=10)
    plt.tight_layout(rect=(0, 0.07 if legend else 0, 1, 1)); plt.show()


R_disp = upright_rotation(ex["V"], ex["F"], ex["labels"])
imgs = render_face_colors(ex["V"], ex["F"], PALETTE[ex["labels"]], R_disp=R_disp)
show_row(imgs, titles=["front", "side", "back", "side"], suptitle=ex["name"] + " — ground truth", legend=True)
```

One example from each data source, so we can see the variety of body shapes and poses:

```python
picks = {}
for p in train_files + test_files:
    picks.setdefault(source(p), p)
imgs, titles = [], []
for src, p in picks.items():
    e = load_example(p)
    imgs.append(render_face_colors(e["V"], e["F"], PALETTE[e["labels"]], n_views=1,
                                   R_disp=upright_rotation(e["V"], e["F"], e["labels"]))[0])
    titles.append(src)
show_row(imgs, titles, legend=True)
```

## 5. The orientation problem

Let's see which coordinate axis is the "long" (head-to-feet) direction of each mesh **in the original files**,
by measuring the bounding-box extent along x, y, z.

```python
stats = defaultdict(list)
for p in train_files + test_files:
    e = load_example(p)
    ext = e["V"].max(0).values - e["V"].min(0).values
    split = "test" if p in test_files else "train"
    stats[(split, e["source"])].append((ext / ext.max()).numpy())

print(f"{'split':6s} {'source':15s} {'n':>4s}   mean extent (x, y, z), longest = 1")
for (split, src), v in stats.items():
    print(f"{split:6s} {src:15s} {len(v):4d}   {np.round(np.mean(v, 0), 2)}")
```

The "up" axis differs between sources — y for FAUST, x for SCAPE, a mix for MIT and SHREC — and the
**test set (SHREC) is oriented differently from everything in training**.

A network that looks at raw `(x, y, z)` coordinates would learn "the head is at large y" and fail on the test set.
We could fight this with heavy random-rotation augmentation, but there is a more elegant fix:

> Describe each point by properties of the **surface itself** — properties that do not change when you
> rotate, translate, or even **bend** the mesh (a person raising an arm is still the same surface).
> Such properties are called **intrinsic**.

The standard tool for intrinsic geometry is the **Laplace–Beltrami operator**.

## 6. The Laplace–Beltrami operator and its eigenfunctions

On a flat grid the Laplacian $\Delta f$ measures how much a value differs from the average of its neighbors.
The **Laplace–Beltrami operator** is the same idea on a curved surface. On a triangle mesh it is discretized as the
**cotangent Laplacian**:

$$ (L f)_i = \sum_{j \in N(i)} w_{ij}\,(f_i - f_j), \qquad w_{ij} = \tfrac12(\cot\alpha_{ij} + \cot\beta_{ij}) $$

where $\alpha_{ij}, \beta_{ij}$ are the two angles opposite the edge $(i,j)$. Each vertex also gets a **mass**
$M_i$ = one third of the area of its adjacent triangles (the "lumped mass matrix").

Only angles and areas appear, and those don't change under rotation, so $L$ is rotation-invariant;
it also changes very little when the body bends, because bending barely stretches the surface.

**Eigenfunctions.** Solving the generalized eigenproblem

$$ L\,\phi_k = \lambda_k\, M\, \phi_k, \qquad 0 = \lambda_0 \le \lambda_1 \le \dots $$

gives functions $\phi_k$ on the surface that are the mesh's "vibration modes" (a Fourier basis for the surface):
small $\lambda$ = smooth, large-scale patterns; large $\lambda$ = fine detail.
With only ~750 vertices we can just do a dense eigendecomposition. We keep the first `K_EIG = 128` modes.

Implementation trick: with $\psi = M^{1/2}\phi$ the problem becomes the ordinary symmetric eigenproblem
$M^{-1/2} L M^{-1/2}\,\psi = \lambda\,\psi$, which `torch.linalg.eigh` solves directly.

```python
K_EIG = 128


def cotan_laplacian(V, Fc):
    # Dense cotangent Laplacian L (n x n) and lumped vertex mass M (n,)
    n = len(V)
    W = torch.zeros(n, n, dtype=torch.float64)
    for k in range(3):                                   # angle at vertex o, opposite edge (i, j)
        i, j, o = Fc[:, k], Fc[:, (k + 1) % 3], Fc[:, (k + 2) % 3]
        u, v = V[i] - V[o], V[j] - V[o]
        cot = (u * v).sum(1) / torch.linalg.cross(u, v).norm(dim=1).clamp_min(1e-12)
        W.index_put_((i, j), 0.5 * cot, accumulate=True)
        W.index_put_((j, i), 0.5 * cot, accumulate=True)
    L = torch.diag(W.sum(1)) - W

    M = torch.zeros(n, dtype=torch.float64)
    M.index_add_(0, Fc.reshape(-1), (face_areas(V, Fc) / 3).repeat_interleave(3))
    return L, M.clamp_min(1e-12)


def laplacian_eigen(V, Fc, k=K_EIG):
    L, M = cotan_laplacian(V, Fc)
    Minv_sqrt = M.rsqrt()
    A = Minv_sqrt[:, None] * L * Minv_sqrt[None, :]
    evals, psi = torch.linalg.eigh(A)
    evals, psi = evals[:k].clamp_min(0), psi[:, :k]     # clamp tiny negative round-off of lambda_0
    evecs = Minv_sqrt[:, None] * psi                     # back to phi: orthonormal w.r.t. M
    return evals, evecs, M


t0 = time.time()
evals, evecs, mass = laplacian_eigen(ex["V"], ex["F"])
print(f"eigendecomposition took {time.time() - t0:.3f}s")
print("first eigenvalues:", evals[:6].numpy().round(2))
print("check  phi^T M phi = I :", torch.allclose(evecs.T @ (mass[:, None] * evecs), torch.eye(K_EIG, dtype=torch.float64), atol=1e-6))
```

**What do the eigenfunctions look like?** We color the mesh by $\phi_1, \phi_2, \phi_3, \phi_{10}$
(blue = negative, red = positive). The low modes vary smoothly along the body's main limbs, so they already
"know" about body structure without any coordinates. Higher modes oscillate faster.

(Rendering a per-vertex quantity: we average the 3 vertex values of each face and color the face.)

```python
def scalar_to_face_rgb(values_per_vertex, Fc, cmap="coolwarm"):
    v = values_per_vertex[Fc].mean(1).float()
    v = (v - v.min()) / (v.max() - v.min() + 1e-12)
    return torch.tensor(plt.get_cmap(cmap)(v.numpy())[:, :3])

imgs = [render_face_colors(ex["V"], ex["F"], scalar_to_face_rgb(evecs[:, k], ex["F"]), n_views=1, R_disp=R_disp)[0]
        for k in [1, 2, 3, 10]]
show_row(imgs, titles=[f"eigenfunction phi_{k}" for k in [1, 2, 3, 10]])
```

## 7. Heat Kernel Signature (HKS) — the input features

Imagine placing a unit of heat at point $x$ and letting it spread over the surface for time $t$.
**How much heat is still at $x$** depends on the geometry around $x$:

* at the tip of a hand or foot heat can only flow one way → it stays → **high** value;
* in the middle of the torso heat escapes in all directions → **low** value.

Small $t$ describes local shape; large $t$ describes the position within the whole body (tip of a limb vs. center).
Using the eigenbasis this has a closed form:

$$ \mathrm{HKS}(x, t) = \sum_k e^{-\lambda_k t}\, \phi_k(x)^2 $$

We evaluate it at 16 logarithmically spaced times, giving a **16-dimensional descriptor per vertex**
that is rotation-invariant and almost pose-invariant. This is the input of our network.

Time range: the classical choice is $t \in [\,4\ln 10/\lambda_{\max},\; 4\ln 10/\lambda_1\,]$.
Since all meshes have unit area their eigenvalues are comparable, so we use one shared range computed from
the median eigenvalues over the training set (computed in the next section).

We also normalize each time slice by $\sum_k e^{-\lambda_k t}$ (the "scaled HKS"), and take a log, so
all 16 channels live on a similar numeric scale.

```python
N_HKS = 16


def hks(evals, evecs, times):
    # evals (k,), evecs (n,k), times (T,) -> (n,T)
    decay = torch.exp(-evals[:, None] * times[None, :])           # (k,T)
    h = (evecs ** 2) @ decay                                     # (n,T)
    return torch.log(h / decay.sum(0, keepdim=True) + 1e-12)
```

## 8. Preprocessing the whole dataset

For every mesh we compute once and cache:

* `evals (K,)`, `evecs (n,K)`, `mass (n,)` — the spectral operators, reused by the network every forward pass,
* `F (f,3)` — triangles (to turn vertex predictions into face predictions),
* `labels (f,)` — ground-truth face labels.

Then we choose the HKS time range from the training eigenvalues, compute HKS features, and **standardize**
them (subtract mean, divide by std of the training set) — a standard step that makes optimization easier.

This takes about half a minute.

```python
def preprocess(paths):
    data = []
    for p in paths:
        e = load_example(p)
        e["evals"], e["evecs"], e["mass"] = laplacian_eigen(e["V"], e["F"])
        data.append(e)
    return data

t0 = time.time()
train_data = preprocess(train_files)
test_data = preprocess(test_files)
print(f"preprocessed {len(train_data)} + {len(test_data)} meshes in {time.time() - t0:.1f}s")

lam1 = np.median([d["evals"][1].item() for d in train_data])
lamK = np.median([d["evals"][-1].item() for d in train_data])
HKS_TIMES = torch.logspace(math.log10(4 * math.log(10) / lamK), math.log10(4 * math.log(10) / lam1), N_HKS,
                           dtype=torch.float64)
print(f"median lambda_1 = {lam1:.1f}, lambda_K = {lamK:.0f}")
print("HKS times:", HKS_TIMES.numpy().round(4))

for d in train_data + test_data:
    d["hks"] = hks(d["evals"], d["evecs"], HKS_TIMES)

all_train_hks = torch.cat([d["hks"] for d in train_data])
feat_mean, feat_std = all_train_hks.mean(0), all_train_hks.std(0)


def to_device(d):
    # float32 GPU tensors used during training
    return {
        "x": ((d["hks"] - feat_mean) / feat_std).float().to(device),
        "evals": d["evals"].float().to(device),
        "evecs": d["evecs"].float().to(device),
        "mass": d["mass"].float().to(device),
        "F": d["F"].to(device),
        "y": d["labels"].to(device),
    }

train_dev = [to_device(d) for d in train_data]
test_dev = [to_device(d) for d in test_data]
```

Let's look at a few HKS channels on the example mesh. Small $t$ (left) highlights local curvature details;
large $t$ (right) clearly separates the extremities (hands, feet, head = hot) from the torso (cold).
Notice how the large-$t$ picture already resembles a coarse segmentation.

```python
d0 = train_data[0]
ts = [0, 5, 10, 15]
imgs = [render_face_colors(d0["V"], d0["F"], scalar_to_face_rgb(d0["hks"][:, t], d0["F"], "inferno"),
                           n_views=1, R_disp=R_disp)[0] for t in ts]
show_row(imgs, titles=[f"HKS t={HKS_TIMES[t].item():.4f}" for t in ts])
```

## 9. The model: a small DiffusionNet

HKS alone cannot tell a hand from a foot (both are "tips"). The network has to combine information from
**far-away parts of the surface** — e.g. "this tip is on the same limb as a long thin part attached high up on the torso".

Ordinary graph neural networks only exchange information between neighboring vertices, so they would need
many layers to see far. **DiffusionNet** (Sharp et al., 2022) instead uses **learned heat diffusion**:

**Diffusion layer.** For a feature channel $x$ (one number per vertex), diffusing for time $t$ is, in the eigenbasis,

$$ \mathrm{diffuse}_t(x) = \Phi\; \big(e^{-\lambda t} \odot \Phi^\top M x\big) $$

1. $\Phi^\top M x$ — project the signal onto the eigenfunctions (spectral coefficients),
2. multiply coefficient $k$ by $e^{-\lambda_k t}$ — high frequencies decay faster,
3. $\Phi(\cdot)$ — back to vertices.

Each channel has its **own learnable $t$**: small $t$ = local smoothing, large $t$ = information spreads across
the whole body in a single layer. Because everything is built from $L$, the layer is rotation-invariant too.

**Diffusion block.** `x → [x, diffuse(x)] → MLP → + x` (a residual connection, which makes deep networks
easier to train). We stack 4 blocks of width 128.

**Output.** The network predicts 8 scores (logits) per **vertex**; a **face** score is the average of its 3
vertices. That matches our face labels.

*(The full DiffusionNet also adds "spatial gradient" features that encode directions; we skip them to keep
the code short.)*

```python
class SpectralDiffusion(nn.Module):
    # Per-channel heat diffusion with a learnable time t >= 0
    def __init__(self, channels):
        super().__init__()
        self.t = nn.Parameter(torch.zeros(channels))

    def forward(self, x, evals, evecs, mass):
        coefs = evecs.T @ (mass[:, None] * x)                          # (K, C) spectral coefficients
        decay = torch.exp(-evals[:, None] * self.t.abs()[None, :])     # (K, C)
        return evecs @ (decay * coefs)                                # (n, C)


class DiffusionBlock(nn.Module):
    def __init__(self, channels, dropout=0.3):
        super().__init__()
        self.diffusion = SpectralDiffusion(channels)
        self.mlp = nn.Sequential(
            nn.Linear(2 * channels, channels), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(channels, channels),
        )

    def forward(self, x, evals, evecs, mass):
        x_diffused = self.diffusion(x, evals, evecs, mass)
        return x + self.mlp(torch.cat([x, x_diffused], dim=-1))


class MiniDiffusionNet(nn.Module):
    def __init__(self, in_ch, num_classes, width=128, n_blocks=4):
        super().__init__()
        self.inp = nn.Linear(in_ch, width)
        self.blocks = nn.ModuleList(DiffusionBlock(width) for _ in range(n_blocks))
        self.out = nn.Linear(width, num_classes)

    def forward(self, x, evals, evecs, mass, faces):
        h = self.inp(x)
        for blk in self.blocks:
            h = blk(h, evals, evecs, mass)
        vertex_logits = self.out(h)                 # (n, classes)
        return vertex_logits[faces].mean(1)         # (f, classes) — average over the 3 corners


model = MiniDiffusionNet(N_HKS, NUM_CLASSES).to(device)
print(model)
print("parameters:", sum(p.numel() for p in model.parameters()))
```

## 10. Training

* **Loss:** cross-entropy between face logits and face labels.
* **Optimizer:** AdamW (Adam with weight decay), learning rate 1e-3 decayed with a cosine schedule.
* **One mesh per step.** Meshes have different numbers of vertices, so the simplest batching is one mesh at a time;
  with ~750 vertices each step takes a few milliseconds.
* **Augmentation:** a tiny amount of Gaussian noise on the input features, so the network doesn't over-rely
  on exact HKS values. (No rotation augmentation needed — the features are already rotation-invariant.)
* We evaluate on the test set every few epochs, only to monitor progress.

**About the test set:** we follow the MeshCNN protocol (train on `train/`, report on `test/`). For hyperparameter
tuning in a real project you'd hold out part of the training set as a validation split instead.

```python
def run_epoch(model, data, optimizer=None, noise=0.0):
    train = optimizer is not None
    model.train(train)
    total_loss, correct, count = 0.0, 0, 0
    order = random.sample(range(len(data)), len(data)) if train else range(len(data))
    with torch.set_grad_enabled(train):
        for i in order:
            d = data[i]
            x = d["x"] + noise * torch.randn_like(d["x"]) if train else d["x"]
            logits = model(x, d["evals"], d["evecs"], d["mass"], d["F"])
            loss = F.cross_entropy(logits, d["y"])
            if train:
                optimizer.zero_grad(); loss.backward(); optimizer.step()
            total_loss += loss.item() * len(d["y"])
            correct += (logits.argmax(1) == d["y"]).sum().item()
            count += len(d["y"])
    return total_loss / count, correct / count


EPOCHS = 60
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
history = defaultdict(list)

t0 = time.time()
for epoch in range(1, EPOCHS + 1):
    tr_loss, tr_acc = run_epoch(model, train_dev, optimizer, noise=0.05)
    scheduler.step()
    history["epoch"].append(epoch); history["train_loss"].append(tr_loss); history["train_acc"].append(tr_acc)
    if epoch % 5 == 0 or epoch == 1:
        te_loss, te_acc = run_epoch(model, test_dev)
        history["test_epoch"].append(epoch); history["test_loss"].append(te_loss); history["test_acc"].append(te_acc)
        print(f"epoch {epoch:3d} | train loss {tr_loss:.3f} acc {tr_acc:.3f} | "
              f"test loss {te_loss:.3f} acc {te_acc:.3f} | {time.time() - t0:.0f}s")
```

### Learning curves

If the training curve keeps going down while the test curve goes up we are **overfitting**;
if both are still dropping at the end, more epochs would help.

```python
fig, (a1, a2) = plt.subplots(1, 2, figsize=(11, 3.5))
a1.plot(history["epoch"], history["train_loss"], label="train")
a1.plot(history["test_epoch"], history["test_loss"], "o-", label="test")
a1.set_xlabel("epoch"); a1.set_ylabel("cross-entropy"); a1.set_title("loss"); a1.legend()
a2.plot(history["epoch"], history["train_acc"], label="train")
a2.plot(history["test_epoch"], history["test_acc"], "o-", label="test")
a2.set_xlabel("epoch"); a2.set_ylabel("face accuracy"); a2.set_title("accuracy"); a2.legend()
plt.tight_layout(); plt.show()
```

## 11. Evaluation

Three ways of looking at the result on the test set:

* **Face accuracy** — fraction of triangles labeled correctly. Simple, but big parts (torso) dominate it.
* **Area-weighted accuracy** — fraction of *surface area* labeled correctly, independent of how finely
  each region happens to be triangulated.
* **Per-class IoU** (intersection over union) — for each class, $|\text{pred} \cap \text{gt}| / |\text{pred} \cup \text{gt}|$.
  Small parts (hands, feet) get equal weight here, so this reveals weaknesses that accuracy hides.
  **mIoU** is their mean.
* **Confusion matrix** — row = true class, column = predicted class; off-diagonal cells show *which* parts get mixed up.

```python
@torch.no_grad()
def predict(model, d):
    model.eval()
    return model(d["x"], d["evals"], d["evecs"], d["mass"], d["F"]).argmax(1).cpu()

conf = torch.zeros(NUM_CLASSES, NUM_CLASSES, dtype=torch.long)
area_correct = area_total = 0.0
for d, dd in zip(test_data, test_dev):
    pred = predict(model, dd)
    conf += torch.bincount(d["labels"] * NUM_CLASSES + pred, minlength=NUM_CLASSES ** 2).reshape(NUM_CLASSES, NUM_CLASSES)
    A = face_areas(d["V"], d["F"])
    area_correct += A[pred == d["labels"]].sum().item(); area_total += A.sum().item()

tp = conf.diag().float()
iou = tp / (conf.sum(0) + conf.sum(1) - conf.diag()).float()
print(f"face accuracy:          {tp.sum().item() / conf.sum().item():.3f}")
print(f"area-weighted accuracy: {area_correct / area_total:.3f}")
print(f"mean IoU:               {iou.mean().item():.3f}")
for n, v in zip(CLASS_NAMES, iou):
    print(f"   IoU {n:10s} {v:.3f}")

conf_norm = conf.float() / conf.sum(1, keepdim=True)
plt.figure(figsize=(5.5, 4.5))
plt.imshow(conf_norm, cmap="Blues", vmin=0, vmax=1)
for i in range(NUM_CLASSES):
    for j in range(NUM_CLASSES):
        plt.text(j, i, f"{conf_norm[i, j]:.2f}", ha="center", va="center", fontsize=7,
                 color="white" if conf_norm[i, j] > 0.5 else "black")
plt.xticks(range(NUM_CLASSES), CLASS_NAMES, rotation=45, ha="right"); plt.yticks(range(NUM_CLASSES), CLASS_NAMES)
plt.xlabel("predicted"); plt.ylabel("true"); plt.title("confusion matrix (row-normalized)")
plt.colorbar(); plt.tight_layout(); plt.show()
```

## 12. Looking at the predictions

For a few test meshes: **ground truth**, **prediction**, and an **error map** (red = wrong triangle).
Errors usually concentrate on **part boundaries** — which are somewhat arbitrary even in the human-made labels —
so a boundary error is far less serious than, say, a hand labeled as a foot.

```python
for d, dd in list(zip(test_data, test_dev))[:4]:
    pred = predict(model, dd)
    Rd = upright_rotation(d["V"], d["F"], d["labels"])
    err = torch.where((pred == d["labels"])[:, None], torch.tensor([0.85, 0.85, 0.85]), torch.tensor([0.9, 0.1, 0.1]))
    imgs = [render_face_colors(d["V"], d["F"], c, n_views=1, R_disp=Rd)[0]
            for c in (PALETTE[d["labels"]], PALETTE[pred], err)]
    acc = (pred == d["labels"]).float().mean().item()
    show_row(imgs, ["ground truth", "prediction", "errors"], suptitle=f"{d['name']}   accuracy {acc:.3f}")
```

## 14. Summary and next steps

**What we did**

* Rebuilt MeshCNN's edge order to convert per-edge labels into per-face labels.
* Saw that the meshes come in inconsistent orientations, and avoided the problem with **intrinsic features**:
  cotangent Laplacian → eigenfunctions → Heat Kernel Signature.
* Trained a compact **DiffusionNet** whose layers spread information across the surface via learned heat diffusion.

**Ideas to improve**

* **Spatial gradient features** (the missing piece of full DiffusionNet) — they let the network know directions
  along the surface and help separate parts that look alike intrinsically (forearm vs. upper arm, shin vs. thigh).
* **Mix in xyz with random-rotation augmentation** as extra input channels.
* **More epochs / wider network / dropout tuning**, using a proper validation split.
* **Soft labels** (`sseg/*.seseg`) as training targets (label smoothing near boundaries), and MeshCNN's
  soft-edge accuracy for a direct comparison with published numbers.
* **Test-time robustness:** try remeshed or noisy versions of the test meshes — intrinsic methods should barely change.
